```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

print("Libraries loaded!")
```

LINEAR ALGEBRA: VECTORS


```
# Vector - a list of numbers representing a point or direction in space (e.g. [x, y, z]); building block of all ML data.
v1 = np.array([2, 3])
v2 = np.array([4, -1])

print("Vector v1:", v1)
print("Vector v2:", v2)

# Vector addition & subtraction - combining vectors component-wise.
print("v1 + v2 =", v1 + v2)
print("v1 - v2 =", v1 - v2)

# Scalar multiplication - stretching/shrinking a vector by a number.
print("3 * v1 =", 3 * v1)

# Vector magnitude (norm) - the "length" of a vector, calculated using Pythagoras' theorem.
magnitude_v1 = np.linalg.norm(v1)
print("Magnitude of v1:", round(magnitude_v1, 2))

# Visualizing vectors as arrows
plt.figure(figsize=(5, 5))
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', label='v1')
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='red', label='v2')
plt.xlim(-5, 6)
plt.ylim(-5, 6)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.title("Vectors v1 and v2")
plt.grid(True)
plt.show()
```

LINEAR ALGEBRA: DOT PRODUCT

```
# Dot Product - multiplies matching elements of two vectors and sums them up; tells you how much two vectors "align" with each other.
dot_product = np.dot(v1, v2)
print("Dot product of v1 and v2:", dot_product)

# Dot product is also used to find the angle between two vectors.
cos_angle = dot_product / (np.linalg.norm(v1) * np.linalg.norm(v2))
angle_deg = np.degrees(np.arccos(cos_angle))
print("Angle between v1 and v2 (degrees):", round(angle_deg, 2))
```

LINEAR ALGEBRA: MATRICES

```
# Matrix - a 2D grid of numbers (rows x columns); used to represent datasets, transformations, and weights in ML models.
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

# Matrix addition
print("A + B:\n", A + B)

# Matrix multiplication - combining two matrices using dot products of rows and columns (core operation in neural networks).
print("A @ B (matrix multiplication):\n", A @ B)

# Transpose - flipping a matrix over its diagonal (rows become columns).
print("Transpose of A:\n", A.T)

# Identity matrix - a special matrix that acts like "1" in multiplication (multiplying by it changes nothing).
identity = np.eye(2)
print("Identity matrix:\n", identity)
print("A @ Identity:\n", A @ identity)

# Determinant - a single number that tells you if a matrix can be "inverted" and how it scales area/volume.
det_A = np.linalg.det(A)
print("Determinant of A:", round(det_A, 2))

# Inverse - the matrix that "undoes" another matrix's transformation (like division, but for matrices).
inv_A = np.linalg.inv(A)
print("Inverse of A:\n", inv_A)
print("A @ Inverse(A) = Identity check:\n", np.round(A @ inv_A))
```

LINEAR ALGEBRA: EIGENVALUES & EIGENVECTORS (Intuition)

```
# Eigenvalues/Eigenvectors - for a matrix, an eigenvector is a special direction that doesn't change when the matrix transforms it,
# it only gets stretched/shrunk, and the eigenvalue tells you by how much. Used heavily in PCA, dimensionality reduction.
M = np.array([[2, 0], [0, 3]])
eigenvalues, eigenvectors = np.linalg.eig(M)

print("Matrix M:\n", M)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# Visual intuition: applying M to its own eigenvector only scales it, doesn't rotate it
ev = eigenvectors[:, 0]
transformed_ev = M @ ev

plt.figure(figsize=(5, 5))
plt.quiver(0, 0, ev[0], ev[1], angles='xy', scale_units='xy', scale=1, color='green', label='Original eigenvector')
plt.quiver(0, 0, transformed_ev[0], transformed_ev[1], angles='xy', scale_units='xy', scale=1, color='orange', label='After transformation (M @ ev)')
plt.xlim(-2, 4)
plt.ylim(-2, 4)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.legend()
plt.title("Eigenvector stays on the same line, just scaled")
plt.grid(True)
plt.show()
```

CALCULUS: DERIVATIVES

```
# Derivative - measures how fast a function's output changes as its input changes; basically the "slope" at a point.
x = sp.Symbol('x')
f = x**2 + 3*x + 5

derivative_f = sp.diff(f, x)
print("Function f(x):", f)
print("Derivative f'(x):", derivative_f)

# Evaluating the derivative (slope) at a specific point, e.g. x = 2
slope_at_2 = derivative_f.subs(x, 2)
print("Slope of f(x) at x=2:", slope_at_2)

# Visualizing function and its tangent line (derivative) at x=2
f_func = sp.lambdify(x, f, 'numpy')
f_prime_func = sp.lambdify(x, derivative_f, 'numpy')

x_vals = np.linspace(-5, 5, 100)
y_vals = f_func(x_vals)

x0 = 2
y0 = f_func(x0)
slope = f_prime_func(x0)
tangent_line = slope * (x_vals - x0) + y0

plt.figure(figsize=(7, 5))
plt.plot(x_vals, y_vals, label='f(x) = x² + 3x + 5')
plt.plot(x_vals, tangent_line, '--', color='red', label=f'Tangent at x={x0} (slope={slope})')
plt.scatter([x0], [y0], color='black', zorder=5)
plt.legend()
plt.title("Function and its Derivative (Tangent Line)")
plt.grid(True)
plt.show()
```

CALCULUS: GRADIENTS

```
# Gradient - for a function with multiple inputs, the gradient is a vector of partial derivatives;
# it points in the direction of steepest increase.
# Gradient descent moves in the opposite direction to minimize error/loss.

x_sym, y_sym = sp.symbols('x y')
g = x_sym**2 + y_sym**2
grad_x = sp.diff(g, x_sym)
grad_y = sp.diff(g, y_sym)

print("Function g(x,y):", g)
print("Gradient: (∂g/∂x, ∂g/∂y) =", (grad_x, grad_y))

# Evaluate gradient at point (2, 3)
grad_at_point = (grad_x.subs({x_sym: 2, y_sym: 3}),
    grad_y.subs({x_sym: 2, y_sym: 3}))

print("Gradient at (2,3):", grad_at_point)

# Convert SymPy values to normal Python numbers
grad_x_value = float(grad_at_point[0])
grad_y_value = float(grad_at_point[1])

# Visualizing the bowl-shaped loss function
X, Y = np.meshgrid(np.linspace(-5, 5, 50),
    np.linspace(-5, 5, 50))
Z = X**2 + Y**2
plt.figure(figsize=(6, 5))

# Contour plot
contour = plt.contour(X, Y, Z, levels=20, cmap='viridis')
plt.clabel(contour, inline=True, fontsize=8)

# Negative gradient direction
plt.quiver(2, 3,-grad_x_value,-grad_y_value,color='red',scale=30,label='Negative Gradient')

# Starting point
plt.scatter([2], [3],color='black', zorder=5,label='Point (2,3)')

plt.legend()
plt.title("Gradient Descent Intuition: g(x,y) = x² + y²")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

CALCULUS: CHAIN RULE (Backprop Intuition)

```
# Chain Rule - if one function is nested inside another (y depends on u, u depends on x), the chain rule lets you find
# how y changes with respect to x by multiplying the individual derivatives together. This is the core math behind
# backpropagation in neural networks, where errors are passed backward layer by layer.
x_sym2 = sp.Symbol('x')
u = 2*x_sym2 + 1        # inner function (like a neuron's weighted input)
y = u**2                 # outer function (like an activation/loss applied on top)

y_composed = y.subs(sp.Symbol('u'), u) if False else (2*x_sym2 + 1)**2  # y as a function of x directly

dy_dx_direct = sp.diff(y_composed, x_sym2)
print("y as function of x:", y_composed)
print("dy/dx (direct):", dy_dx_direct)

# Now verify using chain rule manually: dy/du * du/dx
u_sym = sp.Symbol('u')
y_outer = u_sym**2
dy_du = sp.diff(y_outer, u_sym)          # derivative of outer function
du_dx = sp.diff(u, x_sym2)                # derivative of inner function

dy_du_at_u = dy_du.subs(u_sym, u)         # substitute u back in terms of x
chain_rule_result = sp.simplify(dy_du_at_u * du_dx)

print("dy/du:", dy_du, "| du/dx:", du_dx)
print("Chain rule (dy/du * du/dx):", chain_rule_result)
print("Matches direct derivative:", sp.simplify(chain_rule_result - dy_dx_direct) == 0)
```
