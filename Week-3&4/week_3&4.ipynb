```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("superstore.csv")

print(df.head())
```

```python
def total_sales(data):
    print("Total Sales:", data["Sales"].sum())

total_sales(df)
```

```python
def highest_state(data):
    state = data.groupby("State")["Sales"].sum().idxmax()
    sales = data.groupby("State")["Sales"].sum().max()

    print(state)
    print(sales)

highest_state(df)
```

```python
class SuperStore:

    def __init__(self, dataframe):
        self.df = dataframe

    def total_orders(self):
        print("Orders:", len(self.df))

    def total_sales(self):
        print("Sales:", self.df["Sales"].sum())

    def categories(self):
        print(self.df["Category"].unique())

store = SuperStore(df)

store.total_orders()
store.total_sales()
store.categories()
```
